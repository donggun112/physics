```python
import math
```

수평으로 던진 공의 운동

```python
# 초기속도 입력
v0=float(input())
print("초기속도 : ",v0,"m/s")
```

```python
#초기 각도 입력
angle_degree = float(input())
print("초기 각도 :",angle_degree,"도")
#각도를 radian으로 바꾸기
angle_radian = math.radians(angle_degree)
print("초기 각도 : ",angle_radian,"radian")
```

```python
#x방향의 초기속도
v0x = v0 * math.cos(angle_radian)
print("x방향의 초기속도 : ",v0x,"m/s")
```

```python
#y방향의 초기속도
v0y = v0 * math.sin(angle_radian)
print("y방향의 초기속도 : ",v0y,"m/s")
```

```python
#최고점에 도착하는 시간
t=v0y/9.8
print("최고점에 도착하는 시간 : ",t, t)
```

```python
#수평으로 이동한 최고 거리
R=2*t
print("수평으로 이동한 최고 거리",R,"m")
```

```python
#지면에 닿을때의 Vx, Vy
vx=v0x
vy=v0y
V_t = math.sqrt(vx**2 + vy**2)

print("지면에 닿을 때의 Vx, Vy:", vx, vy, "m/s")
print("지면에 닿을 때의 속도:", V_t, "m/s")
```
