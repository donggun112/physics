```python
import math
```

```python
# 변수 설정
v0 = 100
y0 = 140
theta_rad = math.radians(37)
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)
g = 9.8
```

```python
#시간 구하기
a = 0.5 * g
b = v0
c = -y0

discriminant = b**2 - 4*a*c
t = (-b + math.sqrt(discriminant)) / (2*a)

print("충돌까지 시간: ",t,"m/s")
```

```python
#수평거리
x = v0x * t
print("수평거리 x: ",x,"m")
```

```python
#점 P에 부딪치기 직전의 속도 성분
vx = v0x
vy = v0y - g * t
print("수평성분: vx =", vx, "m/s,")
print("수직성분: vy =", vy, "m/s" )
```

```python
#속도의 크기
v = math.sqrt(vx**2 + vy**2)

print("속도 크기 = ",v, "m/s")
```

```python
#속도의 크기와 수평면과 이루는 각도
v = math.sqrt(vx**2 + vy**2)
angle_rad = math.atan2(vy, vx)
angle_deg = math.degrees(angle_rad)
print("수평면과 이루는 각도°",angle_deg,"도")
```
